```python
# Importing Neccesarry Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import faiss
from google import genai
from sentence_transformers import SentenceTransformer
from langchain_text_splitters import RecursiveCharacterTextSplitter
import yfinance as yf
```

```python
#Importing the Stock Data
ticker = "AAPL"
stock = yf.Ticker(ticker)
```

```python
#Importing the History Data of the Stock
hist_data = stock.history(period="max")
hist_data.head()
```

```python
company_info = stock.info
print(company_info)
```

```python
try:
    new = stock.news
    print(new)
except Exception as e:
    print(f"Could not retrieve stock news at this time: {e}")
```

```python
#Create a Financial Knowledege Base
#Base = hist_data + company_info + new
documents = []
company_text = f"""
Industry: {company_info['industry']}
Sector: {company_info['sector']}
Description: {company_info['longBusinessSummary']}
"""
documents.append(company_text)
print(documents)
```

```python
hist_text = hist_data.to_string()
documents.append(hist_text)
print(documents)
```

```python
# Re-initialize documents list correctly and run splitting
documents = [
    company_text,
    hist_data.to_string()
]

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=0)
chunks = text_splitter.create_documents(documents)
print(f"Successfully created {len(chunks)} chunks!")
```

```python
# Embeddings of Chunks
embedding_model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
text = [doc.page_content for doc in chunks]
embeddings = embedding_model.encode(text)
print(embeddings)
```

### **FAISS Vector Database Indexing**
In this step, we initialize a FAISS Flat L2 index to store our financial knowledge base embeddings, allowing us to perform fast, grounded semantic searches.

```python
# FAISS Vector Database Indexing
embedding_dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(embedding_dimension)

# Adding our generated embeddings to the index
index.add(embeddings)
print(f"Successfully indexed {index.ntotal} vector embeddings into FAISS Vector DB!")
```

### **Defining the Semantic Retrieval Mechanism**
We write a helper function to retrieve the most relevant financial information based on user queries.

```python
# Semantic Search Retrieval Function
def retrieve_financial_context(query, k=5):
    # Embed the query using the same SentenceTransformer model
    query_vector = embedding_model.encode([query])

    # Search the vector database for nearest neighbors
    distances, indices = index.search(query_vector, k)

    # Fetch matching document chunks
    retrieved_chunks = [chunks[idx].page_content for idx in indices[0]]
    return "\n\n---\n\n".join(retrieved_chunks)
```

### **Automated AI Fundamental Financial Analyst Execution**
Now, we combine the FAISS indexing retrieval mechanism with Gemini LLM to answer fundamental stock analysis questions based purely on retrieved data.

```python
# RAG Pipeline: Retrieval + Response Generation
def analyze_stock(query):
    # 1. Retrieve the ground-truth financial facts
    context = retrieve_financial_context(query, k=5)

    # 2. Design a structured prompt guiding Gemini with instructions
    prompt = f"""
    You are a professional Financial Analyst. Your job is to answer the query using only the provided financial data.
    Avoid making up figures or fabricating context that is not present in the reference documents.

    Context:
    {context}

    Query: {query}

    Structure your analysis clearly with highlights, risks, or key metrics where appropriate.
    """

    # 3. Request generation from Gemini
    try:
        response = client.models.generate_content(
            model='gemini-3.5-flash-lite',
            contents=prompt,
        )
        return response.text
    except Exception as e:
        return f"Could not query Gemini: {e}"
```

Let's test the complete pipeline on key analytical topics requested in the capstone assignment (such as business summaries and historical prices).

```python
# Test Query 1: Business Overview
query_1 = f"What are the key business areas, products, and services of {ticker}?"
print(f"Query: {query_1}\n")
print(analyze_stock(query_1))
```
